```python
import os
import pandas as pd
```

```python
raw = "../data/raw/"
```

```python
dp = pd.concat([pd.read_csv(raw + "deepset_train.csv"), pd.read_csv(raw + "deepset_test.csv")])
sg = pd.concat([pd.read_csv(raw + "safeguard_train.csv"), pd.read_csv(raw + "safeguard_test.csv")])
jb = pd.read_csv(raw + "itw_jailbreak.csv")
rg = pd.read_csv(raw + "itw_regular.csv")

d1 = pd.DataFrame({"text": dp["text"], "label": dp["label"].map({0: "safe", 1: "attack"}), "dataset": "deepset"})
d2 = pd.DataFrame({"text": sg["text"], "label": sg["label"].map({0: "safe", 1: "attack"}), "dataset": "safeguard"})
d3 = pd.DataFrame({"text": jb["prompt"], "label": "jailbreak", "dataset": "itw_jailbreak"})
d4 = pd.DataFrame({"text": rg["prompt"], "label": "safe", "dataset": "itw_regular"})

all_df = pd.concat([d1, d2, d3, d4], ignore_index=True)

os.makedirs("../data/processed", exist_ok=True)
all_df.to_csv("../data/processed/merged_v0.csv", index=False)

print(all_df.groupby(["dataset", "label"]).size())
print(len(all_df))
```
