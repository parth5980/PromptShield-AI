```python
sg = load_dataset("xTRam1/safe-guard-prompt-injection")
print(sg)
```

```python
sgdf = sg["train"].to_pandas()

print(sgdf["label"].value_counts())
print()
print(sgdf["text"].str.len().describe())
print()
for lab in [0, 1]:
    print("LABEL", lab)
    for t in sgdf[sgdf["label"] == lab]["text"].head(3):
        print(t[:200])
        print("-----")
```

```python
atk = sgdf[sgdf["label"] == 1].sample(15, random_state=1)
for t in atk["text"]:
    print(t[:150].replace("\n", " "))
    print("-----")
```

```python
sg["train"].to_pandas().to_csv("../data/raw/safeguard_train.csv", index=False)
sg["test"].to_pandas().to_csv("../data/raw/safeguard_test.csv", index=False)
print("saved")
```
